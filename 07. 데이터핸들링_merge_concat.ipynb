```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'이름' : ['재원', '준석', '성훈', '수민', '희영', '소영'],
                     '성별' :['남자', '남자', '남자', '여자', '여자', '여자']})
df1
```

```python
df2 = pd.DataFrame({'이름' : ['재원', '준석', '성훈', '수민', '희영', '소영'],
                     '키' :['180', '190', '176', '165', '170', '166']})
df2
```

# merge : 공통된 컬럼을 기준으로 데이터 프레임을 결합하는 방법 

```python
df3 = pd.merge(df1,df2)
df3
```

```python
df4 = pd.merge(df1,df2, on = '이름', how = 'inner')
df4
```

- inner : 양쪽에 모두 있는 데이터만 결합
- left : 왼쪽 데이터는 모두 유지, 오른쪽의 빈칸이나 결측치를 nan처리
- right : 오른쪽 데이터는 모두 유지
- outer : 양쪽 데이터를 모두 유지

```python
df5 = pd.merge(df1,df2, on = '이름', how = 'left')
df5
```

```python
df6 = pd.merge(df1,df2, on = '이름', how = 'right')
df6
```
