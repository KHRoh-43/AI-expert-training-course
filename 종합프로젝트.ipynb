스마트팩토리 설비 센터 이상 감지 시스템

- 설비의 온도, 진동, 전류 데이터 실시간으로 확인해서 이상상태를 판단
- 조건 : 온도는 80도 이하, 진동은 5.0이하, 전류는 30A 이하
- for, if, while 등 사용
- 단순로직이 아니라, 현업 전문가 다운 프로그램을 구현

```python

while True:
    온도 = float(input("온도"))
    진동 = float(input("진동"))
    전류 = float(input("전류"))


    print("온도 : {} | 진동 : {} | 전류 : {}".format(온도, 진동, 전류))

    if 온도>=80:
        print("고온 감지")

    if 진동>=5.0:
        print("이상 진동 감지")
      
    if 전류>=30:
        print("이상 전류 감지")
    
    sensor_off = input("on / off")

    if sensor_off=="off":
        print("이상 감지 센서를 종료합니다.")
        break

    
```

등급기준
 - 달성률 100이상이면, 목표 달성
 - 달성률 90%이상~100미만, 목표 근접
 - 90 미만, 목표 미달

구현추가 요구사항
 - 달성률을 계산하는 calculate_rate() 함수 필요
 - 생산 등급을 판정하는 judge_production() 함수 필요
 - 각 라인별 목표량, 생산량, 달성률, 등급 출력
 - 전체 평균 달성률 계산
 - lambda함수로 생산량이 많은 순서대로 결정


[출력 예시]

1라인 - 목표: 1000개, 생산: 950개, 달성률: 95.0%, 등급: 근접

2라인 - 목표: 1200개, 생산: 1300개, 달성률: 108.33%, 등급: 목표 달성

3라인 - 목표: 900개, 생산: 720개, 달성률: 80.0%, 등급: 미달

4라인 - 목표: 1100개, 생산: 1080개, 달성률: 98.18%, 등급: 근접

5라인 - 목표: 1000개, 생산: 1150개, 달성률: 115.0%, 등급: 목표 달성

전체 평균 달성률: 99.3%

생산량 높은 순서:

2라인

5라인

4라인

1라인

3라인

```python
lines = [
    {"line": "1라인", "target": 1000, "production": 950},
    {"line": "2라인", "target": 1200, "production": 1300},
    {"line": "3라인", "target": 900, "production": 720},
    {"line": "4라인", "target": 1100, "production": 1080},
    {"line": "5라인", "target": 1000, "production": 1150}
]
```

```python
# 달성률 계산
def calculate_rate(lines):
    success_rate = []
    for i in lines:
        success_rate.append(round(i['production']/i['target']*100, 2))
    return success_rate
```

```python
s_rate = calculate_rate(lines)
s_rate
```

```python
#생산 등극 판정
def judge_production(s):
    p_class = []
    for i in s:
        if i >= 100:
            p_class.append("목표 달성")
        elif i>=90:
            p_class.append("근접")
        else:
            p_class.append("미달")
    return p_class
```

```python
p_class = judge_production(s_rate)
p_class
```

```python
for i in range(len(lines)):
    lines[i].update({
        'rate': s_rate[i],
        'class': p_class[i]
    })

lines
```

```python
#각 라인별 목표량, 생산량, 달성률, 등급 출력
for i in lines:
    print("{}, 목표 : {}개, 생산 : {}개, 달성률 : {}, 등급 : {}".format(i['line'],i['target'],i['production'],i['rate'],i['class']))
```

```python
#전체 평균 달성률
sum = 0 
for i in lines:
    sum = sum + i['rate']
average = sum/len(lines)
print("전체 평균 달성률 : {}".format(round(average,2)))
```

```python
# 생산량이 많은 순서대로 결정
lines.sort(key=lambda x: x['production'], reverse=True)

print('생산량 높은 순서:')
for i in lines:
    print(" ")
    print(i['line'])
    
```
