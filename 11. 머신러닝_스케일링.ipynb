# 1. 데이터 스케일링(Scaling)
 - 데이터 특성(Feature)간의 범위를 통일시켜서 모델이 데이터를 더 잘 이해하고 학습할 수 있도록 하는 방법
 - 머신러닝의 전처리 과정에서 상당히 중요한 방법
 - sklearn 라이브러리를 사용
 - 여러 특성이 다양한 단위와 범위를 가질때 데이터 범위를 일정한 범위로 변환하는 전처리 과정
 - 정규화(normalization), 표준화(Standardization)

1) 정규화
   - 데이터의각 특성의 최소값과 최대값을 이용해서 0 ~ 1 사이의 범위로 조정
   - 최소값은 0, 최대값은 1로 맞추어서 변환
   - 데이터 값들 간의 상대적 거리는그대로 유지하면서,모든 변수가 동일한 척도를 가지게 만드는 방법
   - sklearn : minmaxscaler, robustscaler

2) 표준화
    - 평균이 0이고 표준편차가 1이 되도록 변환 시키는 방법
    - 이 방법은 최소값과 최대값을 제한하지 않음
    - μ는 평균, σ는 표차
    - 각 데이터 변수에서 평균을 뺀 다음 표준편차로 나눈 스케일링 방식
    - sklearn : standardscaler, z-score

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
data = np.array([[10,2.7,3.6],
               [-100,0,1],
               [200,12.5,-0.5]])
```

```python
scaler = StandardScaler()
```

```python
data_Standardized = scaler.fit_transform(data)
data_Standardized
```

```python
from sklearn.preprocessing import RobustScaler
```

# robustscaler : 평균과 표준편차를 사용하지 않음 = > 대신 IQR값을 사용(데이터의 가운데 50% 범위를 의미)
 - 스케일로 변환값 = (원래값 - 중앙값) / IQR
 - 0 ~ 1사이의범위로 바꾸는 정규화 도구가 아니라, 중앙값과 IQR값을 기준으로 값을 재조정하는 스케일링
 - 만약에 0 ~ 1사이의 범위로 바꾸고 싶은 경우라면 MinMaxScaler를 사용
 - MinMaxScaler : 최소값, 최대값/ 범위는 0 ~ 1
 - StandardScaler : 평균, 표준편차 / 범위는 0 ~ 1 아님
 - RobustScaler : 중앙값, IQR / 범위는 0 ~ 1 아님

```python
# robustscaler : 평균과 표준편차를 사용하지 않음 = > 대신 IQR값을 사용(데이터의 가운데 50% 범위를 의미)
scaler = RobustScaler()
```

```python
robust_scaled = scaler.fit_transform(data)
robust_scaled
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
minmax_scaled = scaler.fit_transform(data)
minmax_scaled
```
