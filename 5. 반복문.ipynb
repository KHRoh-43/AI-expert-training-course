```python
time = 0

while(True):
    print("현재 사용량{}".format(time))
    if(time >= 300):
        print("사용중단 : 오늘 사용향 초과! 자제하고 취업준비!!")
        break
    else: 
        time = time + 50
```

```python
#목표금액 모으기
# 1천만원 목표 , 현재 0원, 하루에 15000원씩 저축
#하루가 지날 때마다 저축 금액이 누적
#목표금액 1천만원 이상일때까지 저축 진행, 목표금액을 달성하는데 걸린 총 일수, 총 분
```

```python
money = 0
day = 0
min = 0 
while(True):
    if money<10000000:
        money = money + 15000
        day = day + 1
    else:
        break
min = day*1440
print("1천만원을 모으는데 걸린 총 일수는 {}일이고, 분으로 환산하면 {}분이다.".format(day,min))
```

```python
with open("saving_result","w", encoding="utf-8") as file:
    money = 0
    day = 0
    min = 0 
    while(True):
        if money<10000000:
            money = money + 15000
            day = day + 1
        else:
            break
    min = day*24*60
    file.write("1천만원을 모으는데 걸린 총 일수는 {}일이고, 분으로 환산하면 {}분이다.".format(day,min))
    print("결과가 saving_result.txt 파일로 저장되었습니다.")
```

```python
#while문을 사용 1~10까지의 숫자를 하나씩 확인 
#숫자가 3의 배수인 경우는 출력하지 않고 skip
#3의 배수가 아닌 숫자가 화면에 출력


#시작숫자 : 0
#숫자는 1씩 증가
#3의 배수가 아닌 수만 출력
#continue 사용
```

```python
count = 0

while count<10:
    count = count + 1
    if count%3!=0:
        print(count)
    elif count%3==0:
        continue    
    else:
        break
```
