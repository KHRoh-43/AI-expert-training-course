# 1. 머신러닝을 위한 학습 데이터

1) 정형데이터
 - 규칙적인 형태를 갖춘 데이터
 - 일정한 구조와 형식 존재
 - 엑셀의 표(행, 열), 데이터베이스(DBMS, MySQL)

2) 비정형데이터
 - 일정한 구조와 형식이 없는 데이터
 - 영상, 이미지, 음성 등
 - 데이터 특징 추출(Feature Extraction)을 필요시 사용
 - 자연어 처리, 컴퓨터 비전 등의 기술에 활용


# 2. Scikit-Learn

- 파이썬 머신러닝을 구현하기 위한 필수적인 라이브러리
- 머신러닝 전 과정에서 다양한 기능, 함수를 제공하는 라이브러리


# 3. 머신러닝 과정

1) 데이터 전처리
 - 표준화, 정규화 (스케일링), 원핫 인코딩, 결측치 처리

2) 차원의 축소
 - 데이터의 복잡도 감소
 - PCA, t-SNE

3) Feature Extraction
 - 필요한 특성(징) 선택

4) 모델 학습
 - 데이터 모델에 맞는 학습 방법을 선택
 - 지도 학습(Supervised) vs 비지도 학습(Unsupervised)
 - 지도 학습 : target값(label)을 제시하고 이를 얻기 위한 학습방법
 - Regression모델, classification(분류)모델 (Decision Tree, Random Forest), SVM, KNN//clustering(K-MEANS) 

5) 모델 평가
 - 모델 학습 방법에 맞는 성능 평가 기준
 - 분류, 회귀냐에 따라서 서로 다른 성능 지표를 제공
 - 분류 : Accuracy, Precision, Sensitivity, F1-score(Confusion Matrix)
 - 회귀 : MSE(평균제곱오차), MAE(평균절대값오차), RMSE(), R**2(결정계수) 일반적으로 MSE나 MAE중 하나 + R**2


6) 모델 선택 및 하이퍼파라미터 튜닝
 - 최적 모델 탐색 및 최적 모델을 위한 하이처파라미터를 찾는 기능
 - Grid Search, cross Validation(교차검증)

상관계수 : -1 ~ +1 => -1 ~ 0, 0 ~ 1

[상관관계 분석의 기준]

- 0 ~ 0.2 : 거의 상관없음
- 0.2 ~ 0.4 : 약한 상관관계
- 0.4 ~ 0.6 : 보통 수준이상의 상관관계
- 0.6 ~ 0.8 : 강한 상관관계
- 0.8 ~ 1.0 : 매우 강한 상관관계
