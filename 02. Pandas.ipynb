# Pandas : dataframe 구현을 가능하게 해주는 라이브러리

- pandas는 파이썬에서 데이터를 분석하고 처리하는데 가장 많이 사용하고 강력한 라이브러리
- 엑셀처럼 테이블 형식의 표 데이터를 핸들링이 가능하면 이 개념을 우리는 데이터 프레임(Dataframe) 구조라고 정의
- csv, excel, database, 웹 데이터 (structured data) 등을 수집해서 정리, 가공, 데이터 분석, 데이터 인사이트 도출
- 이때, 주피터노트북이 이 과정에서 가장 좋은 도구로 평가 받고 있음
    - 영상, 이미지, 신호, 주파수 등 (unstructured data) -> json으로 정형데이터화 할 수 있음
    - Pandas에서 정형데이터를 지원
    - 결론적으로, 테이블 형식의구조화된 데이터를 핸들하고 분석할때 최적화된 도구이자 라이브러리
