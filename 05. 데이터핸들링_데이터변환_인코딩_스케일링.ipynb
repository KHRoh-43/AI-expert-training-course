# 값 변환 - map/apply

```python
edu_level = {
    "고졸" : 1,
    "대졸" : 2,
    "석사" : 3,
    "박사" : 4
}
```

```python
df["교육수준"] = df["교육수준"].map(edu_level)
df.head()
```

```python
def 연봉등급(연봉):
    if 연봉 >= 6000:
        return "상위"
    elif 연봉 >= 4500:
        return "중위"
    else:
        return "하위"
```

## apply를 이용한 데이터 적용

```python
df["연봉등급"] = df["연봉"].apply(연봉등급)
```
