AI 데이터분석 프로젝트에서 어려운 지점이 딱 맞아요. **문제정의**는 “무엇을 풀 것인가”를 정하는 단계고, **인사이트 도출**은 “분석 결과가 그래서 어떤 의미인가”를 말로 바꾸는 단계입니다. 모델링보다 이 두 개가 더 어렵고, 실제 프로젝트 완성도도 여기서 많이 갈립니다.

**문제정의**

문제정의는 단순히 “매출 데이터를 분석한다”, “고객 이탈을 예측한다”가 아니라, 아래처럼 구체화하는 과정입니다.

> 누가, 어떤 상황에서, 어떤 의사결정을 더 잘하기 위해, 어떤 데이터를 사용해서, 무엇을 예측/분류/설명할 것인가?

예를 들어 “고객 이탈 예측”이라고 하면 아직 넓습니다.

좋은 문제정의는 이렇게 바뀝니다.

> 최근 6개월 이용 데이터를 바탕으로 다음 달 이탈 가능성이 높은 고객을 예측해서, 마케팅팀이 할인 쿠폰을 보낼 우선순위를 정한다.

이렇게 정의하면 프로젝트 방향이 뚜렷해집니다.

- 대상: 고객
- 목적: 이탈 방지
- 예측 대상: 다음 달 이탈 여부
- 활용 주체: 마케팅팀
- 결과 활용: 쿠폰 발송 우선순위
- 평가 기준: 이탈 고객을 얼마나 잘 잡는가, 즉 recall이나 F1-score가 중요할 수 있음

문제정의할 때는 아래 질문을 순서대로 던지면 좋아요.

1. **이 분석은 누구를 위한 것인가?**
   - 회사, 고객, 관리자, 사용자, 병원, 학교 등

2. **그 사람이 실제로 하고 싶은 결정은 무엇인가?**
   - 누구에게 쿠폰을 보낼까?
   - 어떤 상품을 더 확보할까?
   - 어떤 환자를 우선 관리할까?
   - 어떤 게시글을 먼저 검토할까?

3. **그 결정을 지금은 왜 잘 못하고 있는가?**
   - 데이터가 많아서 보기 어렵다
   - 기준이 사람마다 다르다
   - 사후 대응만 하고 있다
   - 중요한 패턴을 놓치고 있다

4. **AI/데이터분석이 도와줄 수 있는 형태는 무엇인가?**
   - 예측: 미래 값을 맞추기
   - 분류: 카테고리 나누기
   - 추천: 적합한 항목 제안
   - 군집화: 비슷한 그룹 찾기
   - 이상탐지: 특이한 케이스 찾기
   - 설명/EDA: 현상 이해하기

5. **성공 기준은 무엇인가?**
   - 정확도 85% 이상
   - 이탈 고객의 70% 이상 탐지
   - 매출 감소 원인 후보 3개 도출
   - 고객군을 해석 가능한 4개 그룹으로 분류

문제정의는 보통 이 문장으로 정리하면 됩니다.

> 이 프로젝트는 `[대상]`의 `[문제/현상]`을 해결하기 위해, `[데이터]`를 활용하여 `[예측/분류/분석할 것]`을 수행하고, 최종적으로 `[의사결정/활용방안]`에 도움을 주는 것을 목표로 한다.

예시:

> 이 프로젝트는 온라인 쇼핑몰 고객의 이탈 문제를 해결하기 위해, 구매 이력과 방문 데이터를 활용하여 이탈 가능성이 높은 고객을 예측하고, 마케팅팀의 재방문 유도 캠페인 대상 선정에 도움을 주는 것을 목표로 한다.

**인사이트 도출**

인사이트는 단순한 결과 요약이 아닙니다.

예를 들어 이런 건 결과입니다.

> 20대 고객의 구매 횟수가 가장 많았다.

인사이트는 한 단계 더 나아가야 합니다.

> 20대 고객은 구매 횟수는 많지만 평균 구매 금액은 낮기 때문에, 객단가를 높이는 묶음 상품이나 무료배송 기준 금액 전략이 효과적일 수 있다.

즉 인사이트는 보통 이런 구조를 가집니다.

> 발견한 패턴 + 그 의미 + 가능한 행동

분석 결과를 볼 때는 아래 질문을 붙이면 인사이트로 발전하기 쉽습니다.

1. **그래서 뭐가 중요한가?**
   - 단순히 차이가 있는가가 아니라, 의사결정에 영향을 줄 만큼 중요한가?

2. **왜 이런 결과가 나왔을까?**
   - 원인 후보를 생각해보기
   - 단, 상관관계를 인과관계처럼 말하지 않기

3. **누가 영향을 받는가?**
   - 특정 고객군, 상품군, 지역, 시간대, 연령대 등

4. **무엇을 바꾸면 좋아질까?**
   - 마케팅 전략
   - 운영 방식
   - 추천 방식
   - 재고 관리
   - 위험군 관리

5. **추가로 확인해야 할 것은 무엇인가?**
   - 데이터가 더 필요한 부분
   - A/B 테스트가 필요한 부분
   - 외부 요인 확인

인사이트는 아래 형식으로 정리하면 깔끔합니다.

> `[분석 결과]`를 보면, `[패턴]`이 확인된다. 이는 `[가능한 해석]`을 의미할 수 있으며, 따라서 `[실행 방안]`을 고려할 수 있다.

예시:

> 구매 주기가 30일을 넘긴 고객의 이탈률이 급격히 증가하는 패턴이 확인된다. 이는 일정 기간 재방문이 없는 고객이 이탈 위험군으로 전환될 가능성이 높다는 의미일 수 있다. 따라서 마지막 구매 후 25~30일 사이에 재방문 쿠폰이나 개인화 추천 알림을 보내는 전략을 고려할 수 있다.

또 하나 중요한 건, 인사이트에는 **강도**를 조절해야 합니다.

좋지 않은 표현:

> 나이가 많을수록 구매력이 높다.

더 좋은 표현:

> 데이터상 연령대가 높을수록 평균 구매 금액이 증가하는 경향이 관찰된다. 다만 소득 수준, 상품 선호도 같은 변수가 함께 영향을 미쳤을 가능성이 있어 추가 검증이 필요하다.

프로젝트에서 바로 쓸 수 있는 흐름은 이렇게 잡으면 됩니다.

   - “이 데이터로 무엇을 맞힐까?”보다 “이 데이터가 어떤 결정을 도울 수 있을까?”부터 생각하기

   - 타깃 분포
   - 주요 변수 분포

   - 예: “최근 구매일이 오래된 고객일수록 이탈 가능성이 높을 것이다”
   - 예: “배송 지연 경험이 있는 고객은 재구매율이 낮을 것이다”

5. **분석으로 가설 확인**
   - 그룹 비교
   - 시각화
   - 통계량
   - 모델 feature importance
   - SHAP 등 설명 도구

6. **인사이트로 바꾸기**
   - 결과 → 의미 → 행동 제안

7. **한계와 다음 단계 쓰기**
   - 데이터 기간이 짧음
