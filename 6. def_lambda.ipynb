```python
def update_scores(scores):

    new_scores = []

    for s in scores:
        new = s + 5
        new_scores.append(new)

    print(new_scores)

scores = [90, 70, 65, 85, 95, 90, 80, 75, 80, 60]

update_scores(scores)
```

```python
def get_char_count(lyrics, char):
    count = 0
    for i in lyrics:
        if( i == char):
            count = count + 1
    return count        
```

```python
get_char_count(lyrics, '너')
```
