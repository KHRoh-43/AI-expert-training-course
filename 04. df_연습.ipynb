EDA(탐색적 데이터 분석)

- 데이터 셋의 변수들을 확인
- 데이터의 전체적인 모습을확인
   - 데이터 프레임의형태, 걀측치, 이상치 등 데이터 전반에 대한 분석
- 각 변수들의 전체적인 분포현황을 확인 및 분석
- 변수들 간의 경향성을 확인/분석
- 데이터의 시각적 분석 기법을 적용해서 데이터의 인사이트를 도출
- 머신러닝 모델을 사용해서 데이터 인사이트를 발굴하는데 사용이 가능
- 요인, 변수, x값, input, 독립변수, feature (모두 같은 것들)=> 상관관계 분석 및 도출

혼잡 역사는 어디일까? = 혼잡 역사 분석

```python
#승차총승객수 > 50000명 이상이면 혼잡 역으로 분석
df[df['승차총승객수']>50000]
```

```python
station = df[df['승차총승객수']>50000]['역명'].unique

#print("혼잡역사 : \n",station['역명'],"\n혼잡역사 수 : ",len(station.nunique()))
```

```python
station = df[df['승차총승객수']>50000]['역명'].unique()
num = len(station)
print(f"혼잡역사 : {station}\n 혼잡역사 수 : {num}")
```

```python
bsy_st = df[df['승차총승객수'] >= 50000]['역명'].unique()
num = len(bsy_st)
print(f"혼잡역사 : {bsy_st}\n혼잡역사 개수 : {num}")
```
