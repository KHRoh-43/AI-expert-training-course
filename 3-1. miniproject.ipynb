미니 프로젝트 #1

```python
상품1 = input("상품명 입력: ")
상품2 = input("상품명 입력: ")
상품3 = input("상품명 입력: ")
가격1 = int(input("가격 입력: "))
가격2 = int(input("가격 입력: "))
가격3 = int(input("가격 입력: "))
수량1 = int(input("수량 입력: "))
수량2 = int(input("수량 입력: "))
수량3 = int(input("수량 입력: "))

전체가격 = (가격1*수량1)+(가격2*수량2)+(가격3*수량3)

print("%s : %d, %s : %d, %s : %d 전체 가격 : %d" %(상품1, 가격1, 상품2, 가격2, 상품3, 가격3, 전체가격))

```

```python
이름 = input("이름입력")
용돈 = int(input("용돈"))
교통비 = int(input("교통비"))
식비 = int(input("식비"))
간식비 = int(input("간식비"))

남은금액 = 용돈-교통비-식비-간식비

print(f'{이름}의 하루 용돈은'+' {}원 이다. 남은 금액은{}이다.'.format(용돈,남은금액))
```
