```python
import pandas as pd
```

```python
df = pd.read_csv('./data/직원_실습데이터.csv', encoding = 'utf-8-sig')
df.head()
```

```python
df.shape
```

조건 필터링 - 불리언 인덱싱

```python
high_salary = df[df['연봉']>=5000]
print(f"연봉 5000이상 : {len(high_salary)}명")
print(high_salary.head(3))
```

2. 여러 조건 결합 - AND(&), OR(|)

```python
#AND 개발 부서이면서 연봉이 5000이상
개발_연봉 = df[(df['연봉']>=5000)&(df['부서']=='개발')]
print(f"개발 부서이면서 연봉 5000이상 : {len(개발_연봉)}명")
```

```python
#or : 박사 또는경력 15년 이상
박사_경력 = df[(df['경력']>=15)|(df['교육수준']=='박사')]
print(f"박사 또는 경력 15년 이상 : {len(박사_경력)}명")
```

```python
# not : 인사 부서가 아닌 인원(!=)
인사 = df[(df['부서']!='인사')]
print(f"인사 부서가 아닌 인원 : {len(인사)}명")
```

```python
#isin/between-편리한 필터링
```

```python
len(df[df['부서'].isin(['영업','마케팅'])])
```

```python
#경력이 5~10년차
len(df[df['경력'].between(5,10)])
```

# query() - 조건을 문장처럼 

```python
result = df.query("부서 == '개발' and 연봉 >= 4500 and 경력 < 10")
print(f"조건만족 : {len(result)}명")
result.head()
```

## sort_values

```python
#연봉 top3(많은 사람) - 내림차순
high_salary = df.sort_values(by = ['연봉'], ascending = False)
high_salary.head(3)[["사번","부서","연봉"]]
```

```python
#여러기준 : 부서 오름차순 -> 같은 부서 안에서 많은 사람부터 정렬

for i in df['부서'].unique():
    print(df[df['부서']==i].sort_values(by = '연봉', ascending = False).head(3))
    print("\n")
    
```

```python
print(df.sort_values(by=['부서', '연봉'], ascending=[True, False]))
```

```python
print("연봉 TOP 3:\n", df.sort_values("연봉", ascending=False).head(3)[["사번", "부서", "연봉"]])
```

```python
#nlargest/nsmallest(정렬 대신 간편하게 데이터를 핸들링 정렬이 필요할때 사용)

df.nlargest(3,'연봉')[['사번','연봉']]
```

```python
df.nsmallest(3,'연봉')[['사번','연봉']]
```
