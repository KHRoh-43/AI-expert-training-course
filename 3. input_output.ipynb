```python
height = input("키를 입력하세요 : ")
weight = (float(height) - 100) * 0.9
print("적정체중은" + f"{weight:.2f}" + "입니다.")
```

```python
height = float(input("당신의 키를 입력하세요: "))
weight = (height - 100)*0.9
print("당신의 권장 체중은"+str(weight)+"kg입니다.")
```

```python
food = "치킨"
food2 = "맥주"
text = "나는 {}과 {}를 저녁에 먹고 싶다"
print(text.format(food, food2))
```

```python
print("나는 {0}과 {1}을 먹고 싶다. 그런데 우리 학교 기숙사는 {1}이 배달되지 않아서 겁나 짜증난다.".format("치킨","맥주"))
```

```python
#문자열 %s을 이용한 문자 지정 출력 방식
food = "치킨"
print("나는 %s을 먹고 싶다. 저녁에친구 불러서 같이 먹어야지" %food)
```

```python
#정수 출력 : %d
count = 80
print("윗몸일으키기 %d개 하는 사람이다."%count)
```

```python
num = 3.141592
print('pie는 %6.4f입니다.'%num)
```

```python
print('pie는 %5.3f입니다.'%num)
```

```python
player = "손흥민"
track = 100
time = 10.58
print("%s 선수는 %d미터 달리기를 %5.2f초로 달렸다."%(player, track, time))
```
