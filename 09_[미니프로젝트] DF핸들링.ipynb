# 1. 고객, 상품, Q1 주문, Q2 주문 데이터를 불러오시오.

```python

df_cust = pd.read_csv('./data/project2_customers.csv', encoding = 'utf-8-sig')
df_prod = pd.read_csv('./data/project2_products.csv', encoding = 'utf-8-sig')
df_ordq1 = pd.read_csv('./data/project2_orders_q1.csv', encoding = 'utf-8-sig')
df_ordq2 = pd.read_csv('./data/project2_orders_q2.csv', encoding = 'utf-8-sig')
```

# 2. Q1 주문 데이터와 Q2 주문 데이터를 `concat`으로 합치시오.

```python
#2. Q1 주문 데이터와 Q2 주문 데이터를 `concat`으로 합치시오.
q1_q2 = pd.concat([df_ordq1,df_ordq2], ignore_index = True)
q1_q2.head()
```

# 3. 결측치와 중복치를 확인하고 처리하시오.

```python
q1_q2.isnull().sum()
```

```python
q1_q2['할인율'] = q1_q2['할인율'].fillna(0)
```

```python
q1_q2.isnull().sum()
```

```python
q1_q2.duplicated().sum()
```

```python
q1_q2 = q1_q2.drop_duplicates()
q1_q2.duplicated().sum()
```

# 4. 주문일자를 날짜형으로 변환하고 주문월, 요일 변수를 생성하시오.

```python
from datetime import datetime, date
import numpy as np
week = np.array(['월','화','수','목','금','토','일'])
q1_q2["주문일자"] =pd.to_datetime(q1_q2['주문일자'], format='%Y-%m-%d') 
q1_q2['주문월']=q1_q2["주문일자"].dt.month
q1_q2['주문요일']=week[q1_q2["주문일자"].dt.weekday]
q1_q2.head()
```

# 5. 주문 데이터에 고객 정보를 `merge`하시오.

```python
order_cust = pd.merge(q1_q2,df_cust, on = '고객ID', how = 'inner')
order_cust.head()
```

# 6. 주문 데이터에 상품 정보를 `merge`하시오.

```python
order = pd.merge(order_cust,df_prod, on = '상품ID', how = 'inner')
order.head()
```

```python
order.info()
```

# 7. join을 사용하여 주문 데이터에 상품 정보를 인덱스 기준으로 붙이는 예제를 작성하시오.

```python
# join을 사용하여 주문 데이터에 상품 정보를 인덱스 기준으로 붙이기
product_info = df_prod.set_index('상품ID')

order_join = order.set_index('상품ID').join(product_info, how='left', rsuffix='_상품')
order_join.reset_index(inplace=True)
order_join.head()
```

# 9. loc로 VIP 고객 주문만 선택하시오.

# 10. 총매출액, 할인액, 순매출액, 이익, 실매출액 변수를 생성하시오.

```python
#상품ID	수량	할인율	환불여부	 원가	판매가
```

```python
# 총매출액 = 수량*판매가
# 할인액 = 총매출액*할인율
#순매출액 = 총매출액 - 할인액
#이익 = 순매출액 - (원가*수량)
#실매출액 = 순매출액 - 환불량
```

```python
order["총매출액"] = order["판매가"] * order["수량"]
order["할인액"] = order["총매출액"] * order["할인율"]
order["순매출액"] = order["총매출액"] - order["할인액"]
order["이익"] = order["순매출액"] - (order["원가"] * order["수량"])

order["실매출액"] = np.where(
    order["환불여부"] == "환불",
    0,
    order["순매출액"]
)
order.head()
```

# 11. 전자기기 카테고리의 고액 정상 주문을 필터링하고 정렬하시오.

```python
result = order.query("카테고리 == '전자기기' and 환불여부 == '정상' and 판매가>=50000") #50000워 이상의 판매가를 고액으로 정함
elec_order = result.sort_values(by = ['판매가'], ascending = False)
elec_order.head()
```

# 12. 카테고리별 주문건수, 판매수량, 실매출액, 실이익을 집계하시오.

```python
cat_sum = order.groupby('카테고리').agg(
    {
        '주문ID' : 'count',
        '수량' : 'sum', 
        '실매출액' : 'sum', 
        '이익' : 'sum'
    }
).rename(
    columns = {'주문ID' : '주문건수'}).reset_index()

cat_sum
```

# 13. 회원등급을 순서형 인코딩하시오.

```python
order['회원등급'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
order['회원등급라벨'] = le.fit_transform(order['회원등급'])
order.head()
```

# 14. 카테고리, 지역, 결제수단, 판매채널을 원-핫 인코딩하시오.

```python
#부서_원핫 = pd.get_dummies(df['부서'],prefix = '부서', dtype = int)
#부서_원핫.head()

onehot_cols = ["카테고리", "지역", "결제수단", "판매채널"]

order_encoded = pd.get_dummies(
    order,
    columns = onehot_cols,
    dtype = int
)

order_encoded.head()
```

# 15. 매출 관련 수치형 컬럼을 스케일링하시오.

```python
order_encoded.columns
```

```python
# 총매출액, 할인액, 순매출액, 이익, 실매출액
from sklearn.preprocessing import MinMaxScaler

scale_cols = ["총매출액", "할인액", "순매출액", "이익", "실매출액"]
scaler = MinMaxScaler()
df_input_model = order_encoded.copy()
df_input_model[scale_cols] = scaler.fit_transform(df_input_model[scale_cols])

df_input_model.head()
```

# 16. 최종 분석용 데이터와 모델 입력용 데이터를 CSV로 저장하시오.

```python
order.to_csv('./Data/order_analysis.csv', index=False)
df_input_model.to_csv('./Data/order_input_model.csv', index=False)
```
